```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn.model_selection as model_selection
import sklearn.linear_model as linear_model
import sklearn.metrics as metrics
from google.colab import files
uploaded = files.upload()
df= pd.read_csv('vgsales CW-1.csv')
```

Data Explore

```python
# 1. Display First 5 Records

print("\n========== FIRST 5 RECORDS ==========\n")
print(df.head())

# 2. Display Last 5 Records

print("\n========== LAST 5 RECORDS ==========\n")
print(df.tail())

# 3. Dataset Shape

print("\n========== DATASET SHAPE ==========\n")
print(f"Number of Rows    : {df.shape[0]}")
print(f"Number of Columns : {df.shape[1]}")

# 4. Column Names

print("\n========== COLUMN NAMES ==========\n")
print(df.columns.tolist())

# 5. Data Types

print("\n========== DATA TYPES ==========\n")
print(df.dtypes)

# 6. Dataset Information

print("\n========== DATASET INFORMATION ==========\n")
df.info()

# 7. Descriptive Statistics (Numerical Columns)

print("\n========== NUMERICAL SUMMARY ==========\n")
print(df.describe())

# 8. Descriptive Statistics (Categorical Columns)

print("\n========== CATEGORICAL SUMMARY ==========\n")
print(df.describe(include='object'))

# 9. Missing Values Overview

print("\n========== MISSING VALUES ==========\n")
print(df.isnull().sum())

# 10. Missing Value Percentage

print("\n========== MISSING VALUE PERCENTAGE ==========\n")

missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage.sort_values(ascending=False))

# 11. Unique Values Per Column

print("\n========== UNIQUE VALUES ==========\n")
print(df.nunique())

# 12. Unique Platforms

print("\n========== UNIQUE PLATFORMS ==========\n")
print(df['Platform'].unique())

# 13. Unique Genres

print("\n========== UNIQUE GENRES ==========\n")
print(df['Genre'].unique())

# 14. Unique Publishers

print("\n========== TOP 20 PUBLISHERS ==========\n")
print(df['Publisher'].dropna().unique()[:20])
```
